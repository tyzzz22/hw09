```python
# --- 測試一：原始資料 ---
# 固定的兩個活動名稱，以 tuple 保存
activity_names = ("桌遊體驗", "電影欣賞")

# 收到的原始文字名單，保留原值
board_raw = " Amy, Ben, amy, Cora "
movie_raw = " ben, Dora, ERIC, ben "

# 用活動名稱查詢每人費用（dict）
unit_fees = {"桌遊體驗": 120, "電影欣賞": 80}

# 1. 讀取活動名稱，建立兩份保留重複的整理後串列
board_name, movie_name = activity_names
board_clean_list = board_raw.replace(" ", "").lower().split(",")
movie_clean_list = movie_raw.replace(" ", "").lower().split(",")

# 2. 轉成 set 去重並進行集合運算，再排序轉為 list 方便顯示
board_set = set(board_clean_list)
movie_set = set(movie_clean_list)

both_list = sorted(board_set & movie_set)
all_list = sorted(board_set | movie_set)
board_only_list = sorted(board_set - movie_set)
movie_only_list = sorted(movie_set - board_set)

board_unique_list = sorted(board_set)
movie_unique_list = sorted(movie_set)

# 3. 全部參加者排序後，使用切片 [:2] 取得通知名單前兩位
preview_top2 = all_list[:2]

# 4. 使用 dict 保存人數與應收金額（金額保留數值，輸出時再格式化）
board_count = len(board_set)
movie_count = len(movie_set)
both_count = len(both_list)
unique_total_count = len(all_list)
visit_total_count = board_count + movie_count

board_fee = board_count * unit_fees[board_name]
movie_fee = movie_count * unit_fees[movie_name]
total_fee = board_fee + movie_fee

summary = {
    "board_count": board_count,
    "movie_count": movie_count,
    "both_count": both_count,
    "unique_total_count": unique_total_count,
    "visit_total_count": visit_total_count,
    "board_fee": board_fee,
    "movie_fee": movie_fee,
    "total_fee": total_fee,
}

# 5. 輸出完整摘要
print("HW09 活動報名摘要（測試一）")
print(f"活動：{board_name}，每人 {unit_fees[board_name]:.2f} 元")
print(f"活動：{movie_name}，每人 {unit_fees[movie_name]:.2f} 元")
print(f"桌遊整理串列：{board_clean_list}")
print(f"電影整理串列：{movie_clean_list}")
print(f"桌遊去重名單：{board_unique_list}")
print(f"電影去重名單：{movie_unique_list}")
print(f"桌遊人數：{summary['board_count']} 人")
print(f"電影人數：{summary['movie_count']} 人")
print(f"共同參加者：{both_list}")
print(f"共同人數：{summary['both_count']} 人")
print(f"只參加桌遊者：{board_only_list}")
print(f"只參加電影者：{movie_only_list}")
print(f"全部參加者：{all_list}")
print(f"總不重複人數：{summary['unique_total_count']} 人")
print(f"報名人次：{summary['visit_total_count']} 人次")
print(f"通知名單前兩位：{preview_top2}")
print(f"桌遊應收：{summary['board_fee']:.2f} 元")
print(f"電影應收：{summary['movie_fee']:.2f} 元")
print(f"總應收：{summary['total_fee']:.2f} 元")
```
